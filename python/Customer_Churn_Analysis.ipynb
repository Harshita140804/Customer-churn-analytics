```python
df['Total Charges'] = df['Total Charges'].str.strip()
```

```python
df['Total Charges'] = pd.to_numeric(
    df['Total Charges'],
    errors='coerce'
)
```

```python
df['Total Charges'] = df['Total Charges'].fillna(0)
```

```python
df['Total Charges'].isnull().sum()
```

```python
def categorize_tenure(tenure):
    if tenure <= 12:
        return 'Short-term'
    elif tenure <= 36:
        return 'Mid-term'
    else:
        return 'Long-term'
```

```python
df['Customer Segmentation'] = df['Tenure Months'].apply(categorize_tenure)
```

```python
df[['Tenure Months', 'Customer Segmentation']].head(20)
```

```python
print(categorize_tenure(3))
print(categorize_tenure(12))
print(categorize_tenure(13))
print(categorize_tenure(36))
print(categorize_tenure(37))
```
